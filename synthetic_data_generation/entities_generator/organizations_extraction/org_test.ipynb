```python
import pandas as pd

companies_df = pd.read_csv('companies.csv')
companies_df = companies_df.squeeze()
companies_df
```

```python
delete_idx = []
batch_size = 100

for start in range(0, len(companies_df), batch_size):
    batch = companies_df[start:start + batch_size]

    for idx, val in batch.items():
        print(f'id: {idx}, val: {val}')

    raw = input('Delete indices: ').strip()
    if raw:
        delete_idx.extend(map(int, raw.split()))

    cont = input('Continue: (Yes/No): ')
    if cont == 'No':
        print('Stage:', start + len(batch))
        break
```
