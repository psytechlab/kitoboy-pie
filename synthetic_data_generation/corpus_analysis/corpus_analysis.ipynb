```python
import pandas as pd
corpus_csv = pd.read_csv('../data/corpus.csv', sep='|', engine='python', index_col = 'id')  # Или sep='\t' для TSV
```

```python
corpus_csv = corpus_csv.reset_index().drop(columns=['id', 'Unnamed: 0'])
```

```python
corpus_csv.isna().mean()
corpus_csv = corpus_csv.dropna()
```

```python
corpus_csv['text'] = corpus_csv['text'].str.strip() 
```

```python
corpus_csv['text'].sample(10)
```

```python
mask = (corpus_csv['text'].str.len().between(200,500))
mask
```

```python
texts_200_500 = corpus_csv.loc[mask, 'text']
```

```python
import json
res_texts = []
with open('../synthesizer_agent/outputs/synthesized_pii.jsonl', 'r') as f:
    for line in f:
        res_texts.append(json.loads(line))
print(res_texts[0])
```

```python
with open("../synthesizer_agent/outputs/examples_preview.md", "w", encoding="utf-8") as f:
    for i, item in enumerate(res_texts):
        print(
        f'''
        {'-'*100}\n
        EXAMPLE NUMBER {i+1}:\n
        TEXT:\n
        {item['text']}\n
        \n
        LOGIC OF ENTRY:\n
        {item['logic_of_entry']}\n
        \n
        USED ENTITIES:\n
        {item['used_entities']}\n
        \n
        POSITION HINT:\n
        {item['position_hint']}\n
        \n
        SOURCE FRAGMENT:\n
        {item['source_fragment']}
        ''',
        file = f
        )
```
